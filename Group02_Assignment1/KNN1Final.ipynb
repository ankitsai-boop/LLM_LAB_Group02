```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('data.csv')
```

```python
X = df.drop(['id', 'diagnosis', 'Unnamed: 32'], axis=1)
y = df['diagnosis'].map({'B': 0, 'M': 1})
```

```python
y
```

```python
np.random.seed(0)
```

```python
index = np.random.permutation(len(X))
split = int(0.8 * len(X))
```

```python
train_index = index[:split]
test_index = index[split:]
```

```python
X_train = X.iloc[train_index]
X_test = X.iloc[test_index]
y_train = y[train_index]
y_test = y[test_index]
```

```python
mean = X_train.mean(axis=0)
std = X_train.std(axis=0)

X_train = (X_train - mean) / std
X_test = (X_test - mean) / std

X_train = np.round(X_train, 2)
X_test = np.round(X_test, 2)
```

```python
class KNN:
    def __init__(self, k=3, distance='euclidean'):
        self.k = k
        self.distance = distance

    def train(self, X, y):
        # Convert X_train to a NumPy array if it's a DataFrame
        self.X_train = X.to_numpy() if isinstance(X, pd.DataFrame) else X
        self.y_train = y

    def predict(self, X_input):
        # Convert input X to a NumPy array if it's a DataFrame
        X = X_input.to_numpy() if isinstance(X_input, pd.DataFrame) else X_input
        y_pred = []

        for i in range(len(X)):

            # Now both self.X_train and X are NumPy arrays, use NumPy indexing X[i]
            if self.distance == 'euclidean':
                distances = np.sqrt(np.sum((self.X_train - X[i])**2, axis=1))

            elif self.distance == 'manhattan':
                distances = np.sum(np.abs(self.X_train - X[i]), axis=1)

            elif self.distance == 'minkowski':
                distances = np.sum(np.abs(self.X_train - X[i])*3, axis=1)*(1/3)

            elif self.distance == 'cosine':
                dot = np.sum(self.X_train * X[i], axis=1)
                norm = np.linalg.norm(self.X_train, axis=1) * np.linalg.norm(X[i])
                distances = 1 - dot / (norm + 0.000001)

            elif self.distance == 'hamming':
                distances = np.mean(self.X_train != X[i], axis=1)

            index = np.argsort(distances)

            # self.y_train is a Series, so .iloc is still appropriate here
            k_labels = self.y_train.iloc[index[:self.k]]

            prediction = np.bincount(k_labels).argmax()

            y_pred.append(prediction)

        return np.array(y_pred)
```

```python
def accuracy(y_true, y_pred):
    return np.mean(y_true == y_pred)
```

```python
def confusion_matrix(y_true, y_pred):
    tp = np.sum((y_true == 1) & (y_pred == 1))
    tn = np.sum((y_true == 0) & (y_pred == 0))
    fp = np.sum((y_true == 0) & (y_pred == 1))
    fn = np.sum((y_true == 1) & (y_pred == 0))
    return np.array([[tp, fn],
                     [fp, tn]])
```

```python

def precision(y_true, y_pred):
    tp = np.sum((y_true == 1) & (y_pred == 1))
    fp = np.sum((y_true == 0) & (y_pred == 1))
    return tp / (tp + fp + 1e-10)

def recall(y_true, y_pred):
    tp = np.sum((y_true == 1) & (y_pred == 1))
    fn = np.sum((y_true == 1) & (y_pred == 0))
    return tp / (tp + fn + 1e-10)
```

```python
# ==========================
k_values = [3, 4, 9, 20, 47]
distances = ['euclidean','manhattan','minkowski','cosine', 'hamming']

results = {}

best_accuracy = 0
best_k = None
best_distance = None
best_prediction = None
```

```python
for distance in distances:

    acc_list = []

    for k in k_values:

        model = KNN(k=k, distance=distance)
        model.train(X_train, y_train)

        y_pred = model.predict(X_test)

        acc = accuracy(y_test, y_pred)

        acc_list.append(acc)

        if acc > best_accuracy:
            best_accuracy = acc
            best_k = k
            best_distance = distance
            best_prediction = y_pred

    results[distance] = acc_list
```

```python
# ==========================
# Results for every K and every distance
# ==========================

results_table = pd.DataFrame(results, index=k_values)
results_table.index.name = 'K'

# Convert accuracy to percentage
results_table = results_table * 100

display(results_table.style.format("{:.2f}%"))
```

```python
# ==========================
print('Best K        :', best_k)
print('Best Distance :', best_distance)
print('Best Accuracy :', round(best_accuracy, 4))

cm = confusion_matrix(y_test, best_prediction)

print('\nConfusion Matrix')
print(cm)

print('\nPrecision :', round(precision(y_test, best_prediction), 4))
print('Recall    :', round(recall(y_test, best_prediction), 4))
```
